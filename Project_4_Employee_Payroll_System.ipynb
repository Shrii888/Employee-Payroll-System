# Employee Payroll System


```python
import json

try:
    with open("payroll.json", "r") as file:
        employees = json.load(file)

except FileNotFoundError:
    employees = []

print(employees)
```

```python
def save_records():
    with open("payroll.json", "w") as file:
        json.dump(employees, file, indent=4)
```

```python
def add_employee():
    employee_id = int(input("Enter Employee ID: "))

    # Check for duplicate employee ID
    for employee in employees:
        if employee["employee_id"] == employee_id:
            print("Employee ID already exists!")
            return

    name = input("Enter Employee Name: ")
    department = input("Enter Department: ")

    basic_salary = float(input("Enter Basic Salary: "))
    allowances = float(input("Enter Allowances: "))
    deductions = float(input("Enter Deductions: "))

    # Check for negative values
    if basic_salary < 0 or allowances < 0 or deductions < 0:
        print("Salary values cannot be negative!")
        return

    employee = {
        "employee_id": employee_id,
        "name": name,
        "department": department,
        "basic_salary": basic_salary,
        "allowances": allowances,
        "deductions": deductions
    }

    employees.append(employee)

    save_records()

    print("Employee added successfully!")
```

```python
def update_employee(): 
    employee_id = int(input("Enter Employee ID: "))

    for employee in employees:
        if employee["employee_id"] == employee_id:

            print("Current Name:", employee["name"])

            new_name = input("Enter Correct Name: ")

            employee["name"] = new_name

            save_records()

            print("Employee name updated successfully!")

            return

    print("Employee not found!")
```

```python
def calculate_salary():
    employee_id = int(input("Enter Employee ID: "))

    for employee in employees:
        if employee["employee_id"] == employee_id:

            gross_salary = employee["basic_salary"] + employee["allowances"]
            net_salary = gross_salary - employee["deductions"]

            employee["gross_salary"] = gross_salary
            employee["net_salary"] = net_salary

            save_records()

            print("\nSalary Calculated Successfully!")
            print("Gross Salary:", gross_salary)
            print("Net Salary:", net_salary)

            return

    print("Employee not found!")
```

```python
def generate_payslip():
    employee_id = int(input("Enter Employee ID: "))

    for employee in employees:
        if employee["employee_id"] == employee_id:

            gross_salary = employee["basic_salary"] + employee["allowances"]
            net_salary = gross_salary - employee["deductions"]

            employee["gross_salary"] = gross_salary
            employee["net_salary"] = net_salary

            save_records()

            print("\n" + "=" * 40)
            print("           EMPLOYEE PAYSLIP")
            print("=" * 40)

            print("Employee ID   :", employee["employee_id"])
            print("Name          :", employee["name"])
            print("Department    :", employee["department"])

            print("-" * 40)

            print("Basic Salary  :", employee["basic_salary"])
            print("Allowances    :", employee["allowances"])
            print("Gross Salary  :", employee["gross_salary"])
            print("Deductions    :", employee["deductions"])
            print("Net Salary    :", employee["net_salary"])

            print("=" * 40)

            return

    print("Employee not found!")
```

```python
def search_employee():
    employee_id = int(input("Enter Employee ID: "))

    for employee in employees:
        if employee["employee_id"] == employee_id:

            print("\nEmployee Found!")
            print("Employee ID :", employee["employee_id"])
            print("Name        :", employee["name"])
            print("Department  :", employee["department"])
            print("Basic Salary:", employee["basic_salary"])
            print("Allowances  :", employee["allowances"])
            print("Deductions  :", employee["deductions"])

            if "gross_salary" in employee:
                print("Gross Salary:", employee["gross_salary"])
                print("Net Salary  :", employee["net_salary"])

            return

    print("Employee not found!")
```

```python
def main_menu():
    while True:
        print("\n" + "=" * 40)
        print("       EMPLOYEE PAYROLL SYSTEM")
        print("=" * 40)

        print("1. Add Employee")
        print("2. Calculate Salary")
        print("3. Generate Payslip")
        print("4. Search Employee")
        print("5. Save Payroll Data")
        print("6. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            add_employee()

        elif choice == "2":
            calculate_salary()

        elif choice == "3":
            generate_payslip()

        elif choice == "4":
            search_employee()

        elif choice == "5":
            save_records()
            print("Payroll data saved successfully!")

        elif choice == "6":
            save_records()
            print("Thank you for using Employee Payroll System!")
            break

        else:
            print("Invalid choice! Please try again.")
```
